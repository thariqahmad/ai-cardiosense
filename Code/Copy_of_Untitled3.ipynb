```python
import re
import matplotlib.pyplot as plt
```

```python
# Regex untuk cari baris epoch yang mengandung accuracy dan loss
pattern = re.compile(r'accuracy: ([0-9.]+) - loss: ([0-9.]+)')

matches = pattern.findall(log_text)
```

```python
# Deteksi titik awal tiap fold berdasarkan jumlah epoch (misal 100 per fold)
EPOCHS_PER_FOLD = 100
folds_accuracy = []
folds_loss = []

for i in range(0, len(matches), EPOCHS_PER_FOLD):
    fold_data = matches[i:i + EPOCHS_PER_FOLD]
    if len(fold_data) < EPOCHS_PER_FOLD:
        break  # abaikan fold tidak lengkap
    acc = [float(a) for a, _ in fold_data]
    loss = [float(l) for _, l in fold_data]
    folds_accuracy.append(acc)
    folds_loss.append(loss)
```

```python
# -----------------------------------------------
# Step 3: Plot hasilnya (1 gambar per fold, terdiri dari 2 subplot: Accuracy & Loss)
# -----------------------------------------------

for i in range(len(folds_accuracy)):
    plt.figure(figsize=(12, 5))

    # Subplot untuk Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(folds_accuracy[i], label='Accuracy', color='blue')
    plt.title(f'Fold {i+1} - Accuracy per Epoch')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.grid(True)
    plt.legend()

    # Subplot untuk Loss
    plt.subplot(1, 2, 2)
    plt.plot(folds_loss[i], label='Loss', color='red')
    plt.title(f'Fold {i+1} - Loss per Epoch')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.grid(True)
    plt.legend()

    plt.tight_layout()
    plt.show()
```
